> In the digital commerce landscape, customer reviews serve as a direct channel of communication between consumers and businesses, offering candid feedback on products and services. These reviews often contain valuable insights not only about customer satisfaction but also about product performance, usability, and perceived value. However, given the volume and unstructured nature of this data, manually interpreting such feedback at scale is neither efficient nor sustainable. Businesses must adopt analytical methods to classify and understand sentiments across all feedback types—positive, neutral, and negative—to maintain competitiveness, enhance product offerings, and strengthen customer relationships.

<div style="border-left: 6px solid #7c3aed; background: #f5f3ff; padding: 10px 18px; margin: 20px 0 12px 0; border-radius: 0 10px 10px 0;">
<h3 style="color: #4c1d95; margin: 0; font-family: 'Segoe UI', Arial, sans-serif; font-size: 18px; font-weight: 600;">🎯&nbsp; Objective</h3>
</div>

The objective of this project is to build a sentiment analysis system that classifies product reviews into **positive**, **neutral**, or **negative** categories using transformer-based sentence embeddings combined with traditional machine learning classifiers.

Specifically, the goals are to:

1. **Generate meaningful text representations** of customer reviews using a pre-trained Sentence Transformer model (`all-MiniLM-L6-v2`), converting unstructured review text into numerical embeddings suitable for machine learning.
2. **Train and compare multiple classification models** — Random Forest and Gradient Boosting — on these embeddings to identify which approach best predicts review sentiment.
3. **Evaluate model performance** using Accuracy and F1 score (to account for class imbalance among sentiment categories) on both training and test data, checking for overfitting and generalization ability.
4. **Select the best-performing model** for deployment based on test-set performance and consistency between training and test results.

Ultimately, the resulting model aims to help businesses **automatically and scalably interpret large volumes of customer feedback**, enabling faster identification of product strengths, ambiguous or unmet customer expectations, and critical issues — thereby supporting more informed, data-driven decisions in product development, customer service, and brand strategy.

<div style="background: #eef2ff; border-left: 6px solid #6366f1; padding: 14px 20px; border-radius: 0 10px 10px 0; margin: 10px 0 18px 0;">

<p style="margin:0; color:#374151; line-height:1.65;">To convert the unstructured, free-form text in the Product Review column into fixed-length numerical vectors (embeddings) using a pre-trained Sentence Transformer model, so that the review text can be used as input features for machine learning classifiers. The all-MiniLM-L6-v2 model is chosen for this task, as it's a lightweight yet effective pre-trained transformer model well-suited for generating semantic sentence embeddings efficiently. The type() check confirms the data structure of the review column before conversion, and .tolist() converts it into a list format required by the encode() method, with show_progress_bar=True providing visibility into encoding progress across the dataset.</p>

```python
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
```

```python
print(type(data['Product Review']))
```

```python
embedding_matrix = model.encode(
    data['Product Review'].tolist(),
    show_progress_bar=True
)
```

```python
embedding_matrix.shape
```

The model.encode() step then transforms all 1,005 product reviews into dense numerical vectors, capturing the semantic meaning of each review's text (rather than just word-level statistics). The resulting embedding_matrix.shape shows the shape as (1005, 384) — meaning each of the 1,005 reviews has been represented as a 384-dimensional embedding vector, which is the standard output dimensionality of the all-MiniLM-L6-v2 model. This embedding matrix now serves as the numerical feature set (X) that will be used to train and evaluate the sentiment classification models, replacing the need for traditional text vectorization techniques like TF-IDF or Bag-of-Words while capturing richer contextual and semantic relationships between words.</p>

Importantly, stratify=y is used to preserve the original class proportions of Positive, Neutral, and Negative sentiments in both the training and testing sets — this is especially critical given the class imbalance identified during EDA.</p>
</div>

```python
X = embedding_matrix
y = data['Sentiment']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
```

```python
print('Shape of training data:', X_train.shape)
print('Shape of testing data:', X_test.shape)
print('Shape of training labels:', y_train.shape)
print('Shape of testing labels:', y_test.shape)
```

<p style="margin:0; color:#374151; line-height:1.65;">This confirms an 80:20 split of the 1,005 total reviews, with the embedding dimensionality (384) preserved across both sets. The use of stratified splitting ensures that the significant class imbalance observed earlier (Positive ≈ 85%, Neutral ≈ 7–8%, Negative ≈ 7–8%) is proportionally maintained in both the training and testing sets. This is crucial because, without stratification, a random split could result in a test set with very few (or even zero) Neutral or Negative examples, making model evaluation unreliable for those minority classes. With this step complete, the data is now properly structured and ready for model building.</p>

<h2 style="color: #ffffff; margin: 0; font-family: 'Segoe UI', Arial, sans-serif; font-size: 22px; font-weight: 700; letter-spacing: 0.3px;">🏗️&nbsp; Model Building — Transformers + ML</h2>

<div style="border-left: 6px solid #7c3aed; background: #f5f3ff; padding: 10px 18px; margin: 20px 0 12px 0; border-radius: 0 10px 10px 0;">
<h3 style="color: #4c1d95; margin: 0; font-family: 'Segoe UI', Arial, sans-serif; font-size: 18px; font-weight: 600;">📐&nbsp; Model Evaluation Criteria</h3>
</div>

1. **Accuracy** is used to measure the overall correctness of the model by calculating the proportion of total correct predictions across all sentiment classes.
2. **F1 Score** is used to evaluate the balance between precision and recall, providing a more reliable performance metric, especially in the presence of class imbalance among positive, neutral, and negative sentiments.

```python
model_eval_results = pd.DataFrame(columns=['Model', 'Train_Accuracy', 'Train_F1', 'Test_Accuracy', 'Test_F1'])
```

<div style="border-left: 6px solid #7c3aed; background: #f5f3ff; padding: 10px 18px; margin: 20px 0 12px 0; border-radius: 0 10px 10px 0;">
<h3 style="color: #4c1d95; margin: 0; font-family: 'Segoe UI', Arial, sans-serif; font-size: 18px; font-weight: 600;">🌲&nbsp; Random Forest Classifier</h3>
</div>

```python
# Building the model
rf_transformer = RandomForestClassifier(random_state = 42)

# Fitting on train data
rf_transformer.fit(X_train, y_train)
```

```python
from sklearn.metrics import f1_score

# Evaluate the model on training data
y_train_pred_rf = rf_transformer.predict(X_train)
train_accuracy_rf = accuracy_score(y_train, y_train_pred_rf)
train_f1_rf = f1_score(y_train, y_train_pred_rf, average='weighted')

# Evaluate the model on testing data
y_test_pred_rf = rf_transformer.predict(X_test)
test_accuracy_rf = accuracy_score(y_test, y_test_pred_rf)
test_f1_rf = f1_score(y_test, y_test_pred_rf, average='weighted')

# Store results
new_row_rf = pd.DataFrame([{
    'Model': 'RandomForest + Transformer',
    'Train_Accuracy': train_accuracy_rf,
    'Train_F1': train_f1_rf,
    'Test_Accuracy': test_accuracy_rf,
    'Test_F1': test_f1_rf
}])

model_eval_results = pd.concat([model_eval_results, new_row_rf], ignore_index=True)

model_eval_results
```

```python
# Building the model
gb_transformer = GradientBoostingClassifier(random_state = 42)

# Fitting on train data
gb_transformer.fit(X_train, y_train)
```

```python
# Evaluate the model on training data
y_train_pred_gb = gb_transformer.predict(X_train)
train_accuracy_gb = accuracy_score(y_train, y_train_pred_gb)
train_f1_gb = f1_score(y_train, y_train_pred_gb, average='weighted')

# Evaluate the model on testing data
y_test_pred_gb = gb_transformer.predict(X_test)
test_accuracy_gb = accuracy_score(y_test, y_test_pred_gb)
test_f1_gb = f1_score(y_test, y_test_pred_gb, average='weighted')

# Store results
new_row_gb = pd.DataFrame([{
    'Model': 'GradientBoost + Transformer',
    'Train_Accuracy': train_accuracy_gb,
    'Train_F1': train_f1_gb,
    'Test_Accuracy': test_accuracy_gb,
    'Test_F1': test_f1_gb
}])

model_eval_results = pd.concat([model_eval_results, new_row_gb], ignore_index=True)

model_eval_results
```
